```python
# Program for spelling of a digit  (else-if ladder)

a = int(input("Enter a digit : "))
if a == 0:
  print("Zero")
elif a == 1:
  print("One")
elif a == 2:
  print("Two")
elif a == 3:
  print("Three")
elif a == 4:
  print("Four")
elif a == 5:
  print("Five")
elif a == 6:
  print("Six")
elif a == 7:
  print("Seven")
elif a == 8:
  print("Eight")
elif a == 9:
  print("Nine")
else:
  print("Not a valid digit.")
```

```python
# Calculating the grade of exam from mark

mark = int(input("Enter your mark in exam : "))
if mark < 0 or mark > 100 :
    print("Invalid mark")
elif mark >= 80 :
    print("Grade : A+")
elif mark >= 75 :
    print("Grade : A")
elif mark >= 70 :
    print("Grade : A-")
elif mark >= 65 :
    print("Grade : B+")
elif mark >= 60 :
    print("Grade : B")
elif mark >= 55 :
    print("Grade : B-")
elif mark >= 50 :
    print("Grade : C+")
elif mark >= 45 :
    print("Grade : C")
elif mark >= 40 :
    print("Grade : D")
else :
    print("Grade : Fail")
```

### while loop

```python
i = 1  # Initialize the variable i to 1

while i <= 5:  # Condition: As long as i is less than or equal to 5, execute the loop
    print(i)  # Statement indented to indicate it's inside the loop (use 4 space or a tab)
    i = i + 1  # Increment

print("Loop ended")  # Not indented
```
