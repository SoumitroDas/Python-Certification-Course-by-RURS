```python
my_list = [10, 20, 30, 40, 50, 'apple', "banana", 'cherry', True, 4+5j]

print(my_list[0:5])      # elements from index 0 to 4
print(my_list[3:])       # elements from index 3 to the end
print(my_list[:6])       # elements from index 0 to 5
print(my_list[0:9:2])    # elements from index 0 to 8, step 2
print(my_list[-9:-4])    # from index -9 to -5
print(my_list[-9:-1:2])  # from index -9 to -2, step 2
print(my_list[::-1])     # reverse the list
```

### 2D list

```python
A = [[1, 4, 5, 12],
    [-5, 8, 9, 0],
    [-6, 7, 11, 19]]

print("A =", A)
```

```python
print("2nd row =", A[1])
```

```python
print("A23 =", A[1][2])   # 3rd element of 2nd row
print("A13 =", A[0][-1])   # Last element of 1st Row
```

```python
A = [[1, 4, 5, 12],
    [-5, 8, 9, 0],
    [-6, 7, 11, 19]]

for row in A:
  print(row)
```

```python
# Getting a column of the matrix
A = [[1, 4, 5, 12],
    [-5, 8, 9, 0],
    [-6, 7, 11, 19]]

column = []        # empty list/row
for row in A:
  column.append(row[2])

print("3rd column =", column)
```

### Tuple (unchangeable, ordered, allow duplicates)

```python
tp = ('apple', 'banana', 'cherry', 'Rose', 10, True)
print(tp[1])
print(len(tp))
print(type(tp))
```

```python
fruits = ("apple", "banana", "cherry")
(var1, var2, var3) = fruits  # tuple unpacking in variables
print(var1)
print(var2)
print(var3)
```

```python
tp1 = (10, 20, 4+5j, "apple", "banana", "cherry")
(var1, var2, var3, *var4) = tp1
print(var1)
print(var2)
print(var3)
print(var4)
type(var4)
```

```python
my_list = [10, 20, 30, 40, 50, 'apple', "banana", 'cherry', True, 4+5j]
my_tuple = tuple(my_list)               # Converting a list to tuple
print(my_tuple)
```

### Set (unordered/unindexed, unchangeable, mutable(allow changes), don't allow duplicates)

```python
my_set = {4, 3, 1, 2, 3, 4, 5, "hello", "mango"}
print(my_set)
```

```python
ls = [1, 2, 3, 4, 5, 6, 7, 1, 2]
st = set(ls)
print(st)
```

```python
# Set operation
set1 ={1, 2, 3, 4, 5}
set2 ={4, 5, 6, 7}

print(set1 | set2)   #union
print(set1 & set2)   #intersection
print(set1 - set2)   #difference
```

### Dictionary
key : value        paired (ordered, changeable, don't allow duplicates )

```python
mydict =	{
  "name": "Al-Amin",
  "dept": "PSHRD",
  "roll": 2110727128,
  "institute": "RU",
  "age": 23
}                               # Key : Value paired, where key is index name
print(mydict)
```

```python
mydict =	{
  "name": "Al-Amin",
  "dept": "PSHRD",
  "roll": 2110727128,
  "institute": "RU",
  "age": 23
}
print(mydict["roll"])
```

```python
mydict = {
    "name": "Al-Amin",
    "dept": "PSHRD",
    "roll": 2110727128,
    "institute": "RU",
    "age": 23
}

print(f"Name : {mydict['name']} \nDepartment : {mydict['dept']}")
```

```python
td =	{
  "brand": ["Ford", "BMW", "Volvo"],
  "model": "Mustang",
  "year": 1964
}

print(td.get("brand"))
```

```python
td =	{
  "brand": ["Ford", "BMW", "Volvo"],
  "model": "Mustang",
  "year": 1964
}

td["color"] = "white"

print(td)
```
