Q1. Create a class Student with attributes: name, roll, and marks.

• Add a method display() to print the student’s details.

• Create two objects and display their information.

```python
class Student:
     def __init__(self, name, roll, dept):
      self.name = name
      self.roll =roll
      self.dept= dept

     def display(self):
      print("Name: ", self.name, "\nRoll: ", self.roll, "\nDepartment: ", self.dept)
```

```python
S1=Student("Soumitro", "49", "Applied Mathematics")
S1.display()
S2=Student("Tithy", "47", "Food and Nutritional Science")
S2.display()
```

Q2. Write a program where:


• Class A has method display1().


• Class B inherits from Class A and adds method display2().

• Class C inherits from Class B and adds method display3().

• Create an object of Class C and call all three methods.

```python
class A:
  def display1(self):
    print("This is in the vast universe")
class B(A):
  def display2(self):
    print("This is in the cosmos")
class C(B):
  def display3(self):
    print("This is the earth")
```

```python
Earth=C()
Earth.display3()
Earth.display2()
Earth.display1()
```
