#Line Plot Questions (Matplotlib)
1. Basic Line Plot

*   x = [1,2,3,4,5] এবং y = [2,4,6,8,10] দি য়ে একটি simple line plot


```python
x = [1,2,3,4,5]
y = [2,4,6,8,10]
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(x,y, color='r')
```

2. Multiple Line Plot


```python
y1 = [1,2,3,4,5]
y2 = [2,4,6,8,10]
```

```python
plot1 = plt.plot(y1, color='r',label='Line y1')
plot2 = plt.plot(y2, color='b',label='Line y2')
plt.legend(shadow=True)
plt.show()
```

3. Line Style & Marker

```python
Y = [5,7,8,7,6,5,4,3,2]
```

```python
plt.plot(Y, color='r', linestyle='--',marker='o',markersize=5)
```

4. Axis Labels & Title

```python
x = [1,2,3,4,5]
y = [2,3,5,7,11]
plt.plot(x,y, color='r')
plt.xlabel('X Values')
plt.ylabel("Prime Numbers")
plt.title('Prime Line Plot')
plt.show()
```

5. Subplot Line Plots

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))


axes[0].plot(Y, color='r', linestyle='--', marker='o', markersize=5)
axes[0].set_title('Line Style & Marker Plot')



axes[1].plot(x, y, color='r')
axes[1].set_xlabel('X Values')
axes[1].set_ylabel("Prime Numbers")
axes[1].set_title('Prime Line Plot')

 # Adjust layout to prevent overlapping titles/labels
plt.show()
```

#Scatter Plot Questions (Matplotlib)
1. Basic Scatter Plot

```python
x = [5,7,8,7,6,9,5,6,7]
y = [99,86,87,88,100,86,103,87,94]
```

```python
plt.scatter(x,y , color='r', marker='*',s=100)
```

 2. Change Color & Marker  

```python
plt.scatter(x,y , color='g', marker='^',s=100)
```

3. Size of Points

```python
sizes =[20,50,100,200,500,100,250,300,90]
```

```python
plt.scatter(x,y , color='g', marker='^',s=sizes)
```

4. Multiple Scatter Plot

```python
x1=[5,7,8,7,6,9]
y1=[99,86,87,88,100,86]
x2=[6,9,11,12,9,12]
y2=[90,80,85,86,87,85]
```

```python
sz1 =[20,50,100,200,500,100]
sz2 =[200,500,100,250,300,90]
plt.scatter(x1,y1 , color='r', marker='*',s=sz1)
plt.scatter(x2,y2 , color='g', marker='^',s=sz2)
plt.show()
```

5. Scatter with Labels  




```python
plt.scatter(x1,y1 , color='r', marker='*',s=sz1)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study vs Score")
plt.show()
```
