```python
total_confirmed = df['Confirmed'].sum()
print(f"Total Confirmed Cases: {total_confirmed}")
```

```python
total_deaths = df['Deaths'].sum()
print(f"Total Deaths: {total_deaths}")
```

```python
total_recovered =df['Recovered'].sum()
print(f"Total Recovered Cases: {total_recovered}")
```

```python
Catagory = ['confirmed','Deaths','Recovered']
Count = [total_confirmed,total_deaths,total_recovered]
#Create Dataframe
data={'Catagory':Catagory,'Count':Count}
df_Catagory=pd.DataFrame(data)
```

```python
sns.barplot(x='Catagory',y='Count',data=df_Catagory, palette ="viridis")
plt.title("Covid 19 Total Cases by Catagory")
plt.show()
```

```python
df.groupby('Country/Region')['Confirmed'].sum().sort_values(ascending=False).head(10)
```

```python
top10confirmed = df.groupby('Country/Region')['Confirmed'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top10confirmed.index, y=top10confirmed.values, palette='viridis')
plt.title('Top 10 Countries by Confirmed Cases') #to lebel the graph sns is hardes so we used matplotlib.pyplot
plt.xlabel('Country/Region')
plt.ylabel('Confirmed Cases')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df.groupby('WHO Region')['Confirmed'].sum().sort_values(ascending=False).head(3)
```

```python
top3confirmed = df.groupby('WHO Region')['Confirmed'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(12, 6))
ax = sns.barplot(x=top3confirmed.index, y=top3confirmed.values, palette='viridis')
ax.set_title('Top 3 WHO Region by Confirmed Cases')
ax.set_xlabel('Region')
ax.set_ylabel('Confirmed Cases')
plt.xticks(rotation=90, ha='right')
plt.tight_layout()
plt.show()
```
