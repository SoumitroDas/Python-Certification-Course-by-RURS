```python
total_confirmed=df['Confirmed'].sum()
total_deaths=df['Deaths'].sum()
total_recovered=df['Recovered'].sum()
```

```python
print('Total Confirmed:',total_confirmed)
print('Total Deaths:',total_deaths)
print('Total Recovered:',total_recovered)
```

```python
Category=['Confirmed','Deaths','Recovered']
Count=[total_confirmed,total_deaths,total_recovered]
#Create a DataFrame for plotting
data={'Category':Category,'Count':Count}
df_category=pd.DataFrame(data)
```

```python
#Bar Chart plot
sns.barplot(x='Category',y='Count',data=df_category,palette='viridis')
plt.xlabel('Category')
plt.ylabel('Count')
plt.title('Total Cases by Category')
plt.show()
```

```python
# Top 10 Contries Confirmed Case
top10=df.groupby('Country/Region')['Confirmed'].max().sort_values(ascending=False).head(10)
top10
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=top10.index,y=top10.values,palette='viridis')
plt.xlabel('Country')
plt.ylabel('Confirmed Cases')
plt.title('Top 10 Countries with Confirmed Cases')
plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(5,4))
sns.heatmap(df[['Confirmed','Deaths','Recovered']].corr(),annot=True,cmap='Blues')
plt.title('Correlation Between Key Variables')
plt.show()
```
