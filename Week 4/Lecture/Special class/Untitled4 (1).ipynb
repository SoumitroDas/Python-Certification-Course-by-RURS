```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('tips')
```

```python
df.head()
```

```python
sns.histplot(df['total_bill'],kde=True,color='green',bins=10)
```

```python
sns.barplot(x='day', y='total_bill', data=df,color='yellow')
plt.title('Total Bill by Day')
```

```python
sns.barplot(x='day', y='total_bill', data=df,hue='sex')
plt.title('Total Bill by Day')
```

```python
sns.boxplot(x='day', y='total_bill', data=df,hue='sex')
plt.title('Total Bill by Day')
```

```python
sns.violinplot(x='day', y='total_bill', data=df,hue='sex')
plt.title('Total Bill by Day')
```

```python
sns.catplot(x='day', y='total_bill', data=df, kind='point',hue='sex')
plt.title('Total Bill by Day')
```

```python
sns.pairplot(df)
```

```python
  P=sns.PairGrid(df)
  P.map_diag(sns.histplot)
  P.map_upper(sns.scatterplot)
  P.map_lower(sns.kdeplot)
```

```python
sns.lmplot(x='total_bill', y='tip', data=df, hue='sex',col='sex',markers=['o','v'])
sns.set_style('darkgrid')
```

```python
import plotly.express as px
fig = px.pie(df, values='total_bill', names='day', title='Tips Dataset',hole=0.5)
fig.show()
```

```python
C=df.corr(numeric_only=True)
```

```python
sns.heatmap(C,annot=True,cmap='plasma',linewidth=0.2)
```
