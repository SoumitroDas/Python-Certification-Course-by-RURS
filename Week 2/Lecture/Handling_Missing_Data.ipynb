## Handling Missing Data
Missing data is very common in many data analysis applications. pandas has a great ability to deal with the missing data. <br>Let's learn some convenient methods to deal with **missing data in pandas**:<br>

* isnull(), isna(), notnull(), dropna(), fillna(),

```python
import numpy as np
import pandas as pd
```

```python
data_dic = {'A':[1,2,np.nan,4,np.nan],
            'B':[np.nan,np.nan,np.nan,np.nan,np.nan],
            'C':[11,12,13,14,15],
            'E':[16,np.nan,18,19,20]}
df = pd.DataFrame(data_dic)
df
```


**isnull(), isna(), notnull() -- Check for missing data in the dataset!**

```python
df.isnull()
```

```python
df.isnull().sum() #Count all null values in every columns
```

```python
df.isnull().sum().sum()  #Count all null values in total dataset
```

```python
df['A'].isnull()
```

```python
df['A'].isnull().sum()   #Count all null values in a specific column
```

```python
df.isna()     #isnull() and isna() are same
```

```python
df.isna().sum().sum()
```

```python
df.loc[1].isnull().sum()
```

```python
df.notnull()
```

```python
df.shape
```

```python
df.notnull().sum()
```

```python
df.notnull().sum().sum()
```

```python
df
```

```python
# Sum on Column "A", (NaN as 0)
df['A'].sum()
```

&#9758; NaN ignored for mean().

```python
df['A'].mean()
```

```python
df.loc[3].sum()
```

**dropna(), fillna() -- Cleaning / filling the missing data**

```python
df.dropna(axis=1)   #Drop all columns which contains at least 1 null valaue
```

```python
df.dropna(axis=0)   #Drop all rows which contains at least 1 null valaue
```

```python
df
```

```python
df
```

* threshold : int, default None
* thresh = 3 means, it will drop any column that have more than 3 non-NaN values.

```python
df.dropna(thresh=3, axis=1)
```

We can use fillna() to fill in the values.<br>
inplaced = True for permanent change.

```python
df.fillna(value=2)
```

```python
df
```

Let's fill in the values using mean of the column.

```python
df['A'].fillna(value = df['A'].mean())
```

```python
df
```

```python
df.fillna(method='ffill')   #front fill (fill with the value of previous data)
```

```python
df.fillna(method='bfill')    #back fill (fill with the value of next data)
```

```python
# fill with you own given value
df.fillna(0)
```

```python
df2 = df.copy()
df2
```

```python
df
```

```python
#Change the dataset permaently
df2.fillna(0, inplace=True)
```

```python
df2
```
