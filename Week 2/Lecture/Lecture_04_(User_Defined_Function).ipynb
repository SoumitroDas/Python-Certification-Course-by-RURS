```python
'''
This is multi line comment
Used as same as single line comment
'''
def square(a):
  '''
  Return square value
  of a number
  '''
  # doc string (a multi line comment in a function used as a doc string)
  return a*a

x = square(5)
print(x)
print(square.__doc__)   # __doc__ is an attribute (spacial command) to print doc string of a function
```

### lambda function (annonymus function)
lambda arguments : expression

```python
x = lambda a : a * a
x(5)
```

```python
x = lambda a, b: a + b
print(x(10, 20))
```

```python
def fn(a):
  return lambda n : n * a

x = fn(2)      # x = lambda n : n * 2
print(x(11))
```

### map and filter function

```python
marks = [78, 86, 42, 55]
def grade(marks) :
    if marks < 0 or marks > 100 :
        return 'Invalid mark'
    elif marks >= 80 :
        return 'A+'
    elif marks >= 75 :
        return 'A'
    elif marks >= 70 :
        return 'A-'
    elif marks >= 65 :
        return 'B+'
    elif marks  >= 60 :
        return 'B'
    elif marks >= 55 :
        return 'B-'
    elif marks >= 50 :
        return 'C+'
    elif marks >= 45 :
        return 'C'
    elif marks >= 40 :
        return 'D'
    else :
        return 'F'
grades = map(grade, marks)          # map(function, iterable)
                                    # iterable (on which the function will be applied)
print("Exam score : ", marks)
print("Grade : ", next(grades))     # map() is an object. next() used to apply function to one by one element
print("Grade : ", next(grades))
print("Grade : ", next(grades))
```

```python
marks = [78, 14, 86, 42, 31, 55]
def grade(marks) :
    if marks < 0 or marks > 100 :
        return 'Invalid mark'
    elif marks >= 80 :
        return 'A+'
    elif marks >= 75 :
        return 'A'
    elif marks >= 70 :
        return 'A-'
    elif marks >= 65 :
        return 'B+'
    elif marks  >= 60 :
        return 'B'
    elif marks >= 55 :
        return 'B-'
    elif marks >= 50 :
        return 'C+'
    elif marks >= 45 :
        return 'C'
    elif marks >= 40 :
        return 'D'
    else :
        return 'F'
grades = map(grade, marks)          # map(function, iterable)

print("Exam score : ", marks)
print("Grade : ", list(grades))     # list() used to apply function to all element
```

```python
# Without map (one element)
def fn(a):
  return len(a)

x = fn("shaon")
print(x)
```

```python
# With map (all element)
def fn(a):
  return len(a)

a =  ["Apple", "banana", "jackfruit"]
x = list(map(fn,a))

print(x)
```

```python
marks = [78, 14, 86, 42, 31, 55]

def fail(score) :
    return score < 40

result = filter(fail, marks)
print("Failing scores : ", list(result))
```

```python
marks = [78, 14, 86, 42, 31, 55]

def fn(score):
  if score <= 40:
    return True
  else:
    return False

x = list(filter(fn,marks))
print(x)
```

```python
num = [11,20,29,31,42,47,51,60,57,86]
def even(x):
    return x % 2 == 0

even = list(filter(even, num))
print("Even numbers: ", even)
```

```python
num = [11,20,29,31,42,47,51,60,57,86]

even = list(filter(lambda x : x%2==0, num))
print("Even numbers: ", even)
```
